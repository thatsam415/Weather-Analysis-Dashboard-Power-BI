```python
# Logistic Regression vs SVM - complete script
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, roc_curve, precision_recall_curve, auc,
    confusion_matrix, classification_report
)
from sklearn.decomposition import PCA
```

```python
# Load dataset
data = load_breast_cancer(as_frame=True)
X = data.frame.drop(columns=['target'])
y = data.frame['target']
```

```python
# Train / Test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
# Preprocessing pipeline
preprocess = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

pipe_logreg = Pipeline([
    ('prep', preprocess),
    ('clf', LogisticRegression(max_iter=2000, solver='liblinear'))
])
pipe_svm = Pipeline([
    ('prep', preprocess),
    ('clf', SVC(probability=True))
])
```

```python
# Baseline fits
pipe_logreg.fit(X_train, y_train)
pipe_svm.fit(X_train, y_train)
```

```python
# Evaluation helper
def evaluate(pipe, X_test, y_test):
    y_pred = pipe.predict(X_test)
    y_prob = pipe.predict_proba(X_test)[:, 1]
    metrics = {
        'accuracy': accuracy_score(y_test, y_pred),
        'precision': precision_score(y_test, y_pred),
        'recall': recall_score(y_test, y_pred),
        'f1': f1_score(y_test, y_pred),
        'roc_auc': roc_auc_score(y_test, y_prob)
    }
    return metrics, y_pred, y_prob
```

```python
metrics_logreg_base, ypred_logreg, yprob_logreg = evaluate(pipe_logreg, X_test, y_test)
metrics_svm_base, ypred_svm, yprob_svm = evaluate(pipe_svm, X_test, y_test)

print("Baseline LogReg metrics:", metrics_logreg_base)
print("Baseline SVM metrics:", metrics_svm_base)
```

```python
# Hyperparameter search (Randomized)
rkf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
logreg_param_dist = {'clf__C': [0.001, 0.01, 0.1, 1, 10, 100], 'clf__penalty': ['l1','l2'], 'clf__solver': ['liblinear']}
svm_param_dist = {'clf__C': [0.1,1,10,50,100], 'clf__kernel': ['linear','rbf','poly'], 'clf__gamma': ['scale','auto']}

rs_logreg = RandomizedSearchCV(pipe_logreg, logreg_param_dist, n_iter=24, cv=rkf, scoring='roc_auc', n_jobs=-1, random_state=42, verbose=1)
rs_svm = RandomizedSearchCV(pipe_svm, svm_param_dist, n_iter=24, cv=rkf, scoring='roc_auc', n_jobs=-1, random_state=42, verbose=1)

rs_logreg.fit(X_train, y_train)
rs_svm.fit(X_train, y_train)

print("Best LogReg params:", rs_logreg.best_params_, "CV ROC-AUC:", rs_logreg.best_score_)
print("Best SVM params:", rs_svm.best_params_, "CV ROC-AUC:", rs_svm.best_score_)
```

```python
best_logreg = rs_logreg.best_estimator_
best_svm = rs_svm.best_estimator_

metrics_logreg_tuned, ypred_logreg_tuned, yprob_logreg_tuned = evaluate(best_logreg, X_test, y_test)
metrics_svm_tuned, ypred_svm_tuned, yprob_svm_tuned = evaluate(best_svm, X_test, y_test)

print("Tuned LogReg metrics:", metrics_logreg_tuned)
print("Tuned SVM metrics:", metrics_svm_tuned)
```

```python
# ROC plot
fpr_log, tpr_log, _ = roc_curve(y_test, yprob_logreg_tuned)
fpr_svm, tpr_svm, _ = roc_curve(y_test, yprob_svm_tuned)
plt.figure()
plt.plot(fpr_log, tpr_log, label=f"LogReg AUC={metrics_logreg_tuned['roc_auc']:.3f}")
plt.plot(fpr_svm, tpr_svm, label=f"SVM AUC={metrics_svm_tuned['roc_auc']:.3f}")
plt.plot([0,1],[0,1],'--')
plt.xlabel("FPR"); plt.ylabel("TPR"); plt.legend(); plt.title("ROC Curve"); plt.show()
```

```python
# Precision-Recall plot
prec_log, rec_log, _ = precision_recall_curve(y_test, yprob_logreg_tuned)
prec_svm, rec_svm, _ = precision_recall_curve(y_test, yprob_svm_tuned)
plt.figure()
plt.plot(rec_log, prec_log, label=f"LogReg AP={auc(rec_log, prec_log):.3f}")
plt.plot(rec_svm, prec_svm, label=f"SVM AP={auc(rec_svm, prec_svm):.3f}")
plt.xlabel("Recall"); plt.ylabel("Precision"); plt.legend(); plt.title("Precision-Recall"); plt.show()
```

```python
# Coefficients & support vectors
coefs = best_logreg.named_steps['clf'].coef_.ravel()
feat_imp = pd.DataFrame({'feature': X.columns, 'coef': coefs}).sort_values('coef', key=lambda s: s.abs(), ascending=False)
print("Top coefficients:\n", feat_imp.head(10))

svm_clf = best_svm.named_steps['clf']
if hasattr(svm_clf, 'n_support_'):
    print("SVM support vectors per class:", svm_clf.n_support_)
```
